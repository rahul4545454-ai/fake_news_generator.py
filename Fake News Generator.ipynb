```python
import random
subjects=["Sharukh Khan",
          "Ajay Devgan",
          "Nirmal Devi",
          "Karishma Kappor",
          "Vijay Arrora",
          "A Group Of Monkey",
          "Prime Minister Modi",
          "Akshay Kumar",
          "Anil Kappor",
          "Auto Rikshaw Driver From Dehli"]

actions=[
    "launches ",
    "cancel ",
    "enjoy ",
    "eat ",
    "declare war on",
    "love ",
    "kill",
    "order",
    "win",
    "dance with",]

place_or_things=[
    "at red fort"
    "in mumbai local train",
    "inside parmilament"
    "during ipl match",
    "at ganga ghat",
    "at stadium",    
    "at india gate",
    "a plate of samosa",]

#start the headline genration loop

while True:
    subject=random.choice(subjects)
    action=random.choice(actions)
    place_or_thing=random.choice( place_or_things)
    
    headline=f" BREAKING NEWS:{subject} {action} {place_or_thing} "
    print("\n" + headline)
    user_input=input("\nDo you want another headline?(yes/no)").strip().lower()
    if user_input=="no":
        break
        #print good bye message
print("\nThanks for using the fake News Headline Genrator.Have a fun")
    
```
